```python
import datetime
from pandas_datareader import data ,wb
import pandas as pd
import numpy as np 
import matplotlib.pyplot as plt 
import seaborn as sns

import plotly
import yfinance as yf
import matplotlib.pyplot as plt
```

```python
BAC = yf.download("BAC", start="2006-01-01", end="2016-01-01")
C= yf.download("C",start="2006-01-01", end="2016-01-01")
GS= yf.download("GS",start="2006-01-01", end="2016-01-01")
MS= yf.download("MS",start="2006-01-01", end="2016-01-01")
JPM= yf.download("JPM",start="2006-01-01", end="2016-01-01")
WFC= yf.download("WFC",start="2006-01-01", end="2016-01-01")
```

```python
tickers=["BAC","JPM","GS","C","WFC","MS"]
```

```python
bank_stock=pd.concat([BAC,JPM,GS,C,WFC,MS],axis=1,keys=tickers)
bank_stock.columns.names=['Bank ticker','stock info','Banker']
bank_stock.columns = bank_stock.columns.droplevel(2)
bank_stock
```

```python

## the maximum close of the stock in each bank
bank_stock.xs(key="Close",axis=1,level=1).max()


```

```python
returns=pd.DataFrame()
for ticks in tickers:
    returns[ticks +" Return"]=bank_stock[ticks]["Close"].pct_change()

returns
```

```python
## now let's get more into the studying of the 2008 crisis

# isolating data from a key period 2008 (2008-01-01 to 2008-12-31)


crisis_date=bank_stock[bank_stock.index.year==2008]
 ## percentage of change of close price
daily_returns=crisis_date.xs(key="Close",axis=1,level="stock info").pct_change()


# the correlation matrix of daily returns
corr_mat_dReturn=daily_returns.corr()
```

```python

#calculating the annual rolling volatility and annualizing it by multiplying it by the square root of 252

rolling_volatility = returns.rolling(window=30).std() * np.sqrt(252)
## the max and min volatility  for each  bank and ranking them by their mean

for tick in tickers:
    print(tick,rolling_volatility[tick+" Return"].idxmax())
average_volatility=returns.mean()
average_volatility.sort_values(ascending=False)


```

```python
overnight_Gap=bank_stock.xs(key="Open",axis=1,level="stock info").pct_change(periods=1)/(bank_stock.xs(key="Close",axis=1,level="stock info")-1)
# Mean gap
overnight_Gap.mean()
# Standard deviation of gaps
overnight_Gap.std()
# Number of days with a gap down exceeding -2% or -3%
lower_gap=overnight_Gap[overnight_Gap<-3]
lower_gap.value_counts()
# The largest gap down ever.

overnight_Gap.max().max()
```
